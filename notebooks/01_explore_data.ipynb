# 01 - Understanding the Data
This notebook explains what data we have: social media **posts** and the **comments** on them.

```python
import pandas as pd, matplotlib.pyplot as plt
posts = pd.read_csv('../data/posts.csv')
comments = pd.read_csv('../data/comments.csv')
print('Posts:', posts.shape, '| Comments:', comments.shape)
posts.head()
```

**Posts table:** each row is one post (likes, views, content type, etc.).

**Comments table:** each row is one comment, linked to its post through `post_id`.

```python
comments.head(10)
```

## Question 1: How many posts are there on each platform?

```python
posts.platform.value_counts()
```

## Question 2: Which posts have the most likes?

```python
posts.sort_values('likes', ascending=False).head(3)
```

## Question 3: What is the language mix of the comments (English vs Roman Urdu)?

```python
print(comments.language_type.value_counts())
comments.language_type.value_counts().plot(kind='pie', autopct='%1.0f%%', title='Comment language')
plt.ylabel('')
plt.show()
```

## Chart: Average likes by content type

```python
posts.groupby('content_type').likes.mean().sort_values().plot(kind='barh', title='Average likes by content type')
plt.xlabel('Average likes')
plt.show()
```

## My observations

- ...- Reels get the highest average likes (about 2,700), while product photos get the lowest (about 700).
- Most comments (71%) are in Roman Urdu, so the brand should reply to customers in Roman Urdu.
