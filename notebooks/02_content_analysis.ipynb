# 02 - Which Content Attracts Attention?
**Engagement rate = (likes + comments + shares) / views x 100.**
Looking only at likes is misleading, because posts with more views naturally get more likes.

```python
import sys, pandas as pd, matplotlib.pyplot as plt
sys.path.append("..")          # so that engine.py can be imported
import engine as E
posts = pd.read_csv("../data/posts.csv")
comments = pd.read_csv("../data/comments.csv")
posts = E.add_metrics(posts)   # adds the engagement_rate column
```

## Comparing content types

```python
perf = E.content_performance(posts)
perf
```

```python
perf.avg_engagement_rate.sort_values().plot(kind='barh', title='Engagement rate by content type (%)')
plt.show()
```

## Comparing platforms

```python
plat = posts.groupby('platform').engagement_rate.mean().sort_values()
print(plat.round(2))
plat.plot(kind='bar', title='Engagement rate by platform (%)')
plt.show()
```

## Comparing campaigns (Eid vs 11.11 vs 12.12 vs Winter)

```python
camp = posts.groupby('campaign').engagement_rate.mean().sort_values()
print(camp.round(2))
camp.plot(kind='bar', title='Engagement rate by campaign (%)')
plt.show()
```

## Platform x Content type

```python
table = posts.pivot_table(index='content_type', columns='platform', values='engagement_rate', aggfunc='mean').round(2)
table
```

## My conclusions 

1. I noticed that Reels have the highest engagement rate (about 7.6%), while Product Photos have the lowest (about 4.0%).
2. Reels perform best on all three platforms, and especially on TikTok (7.95%), while the difference between platforms and campaigns is small.
3. Therefore the brand should make more Reels (especially for TikTok) and rework its Product Photo posts, for example by adding styling or short video elements.
